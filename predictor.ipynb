```python
data['total_sqft'].unique()
# data me kuch ranges bhi hai--'1133 - 1384', so we gonna fix them
# so iss hyphen ke hisaab se we gonna split and phir dono ka integer value ka mean nikal lenge
# we will do (1133+1384)/2. jahan bhi we get range
```

```python
# ek function hai jo harr ek value lega and uspar ek split maarega.
# if split hoga and 2 value banega then convert both value into float and then calulate mean
# agar split krne ke baad ek hi value mila than we gonna return float
# so agar float ke conversion me problem hua than we are gonna return none

def convertRange(x):

    temp=x.split('-')
    if len(temp)==2:
        return (float(temp[0]) + float(temp[1]))/2
    try:
        return float(x)
    except:
        return None
```

```python
# now apply this function on total_sqft column

data['total_sqft']=data['total_sqft'].apply(convertRange)

# apply() takes one value at a time from a column and applies a function to it.
```

```python
# so remove this OUTLIER this is my function
def remove_outliers_sqft(df):
    df_output=pd.DataFrame()
    for key,subdf in df.groupby('location'): # group by lagane se ek key milta hai,aur uss key ka subdataframe(subdf)milta hai
        m=np.mean(subdf.price_per_sqft) # subdf me price_per_sqft ka mean nikala

        st=np.std(subdf.price_per_sqft) # standardDeviation

        gen_df=subdf[(subdf.price_per_sqft > (m-st)) & (subdf.price_per_sqft <= (m+st))] # using filter on subdf, unhi price ko rakh rha hun jo mean se ek std idhar udar ho, and baki sab prices i am droping
        # iss subdf wo agar m-st and m+st ke under hua i m keeoing them
        df_output=pd.concat([df_output,gen_df],ignore_index=True)
    return df_output
data=remove_outliers_sqft(data)
data.describe()
```

```python
def bhk_outlier_remover(df):
    exclude_indices = np.array([])

    for location, location_df in df.groupby('location'):
        bhk_stats = {}

        for bhk, bhk_df in location_df.groupby('bhk'):
            bhk_stats[bhk] = {
                'mean': np.mean(bhk_df.price_per_sqft),
                'std': np.std(bhk_df.price_per_sqft),
                'count': bhk_df.shape[0]
            }

        for bhk, bhk_df in location_df.groupby('bhk'):
            stats = bhk_stats.get(bhk - 1)

            if stats and stats['count'] > 5:
                exclude_indices = np.append(
                    exclude_indices,
                    bhk_df[bhk_df.price_per_sqft < stats['mean']].index.values
                )

    return df.drop(exclude_indices, axis='index')
```

```python
data=bhk_outlier_remover(data)
```

```python
# BETTER WAY
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

# Define columns
categorical_cols = ['location']
numerical_cols = X_train.columns.drop('location')

# Column-wise preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols),
        ('num', StandardScaler(), numerical_cols)
    ]
)

# Linear Regression pipeline
pipe_lr = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])

# Train
pipe_lr.fit(X_train, y_train)
```

```python
# predict
y_pred_lr=pipe_lr.predict(X_test)
```

```python
r2_score(y_test,y_pred_lr)
```

# RIDGE REGRESSION

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge

categorical_cols = ['location']
numerical_cols = X_train.columns.drop('location')

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols),
        ('num', StandardScaler(), numerical_cols)
    ]
)

# ridge Pipeline
pipe_ridge = Pipeline([
    ('preprocessor', preprocessor),
    ('model', Ridge(alpha=1.0))
])

# train
pipe_ridge.fit(X_train, y_train)

```

```python
# predict
y_pred_ridge = pipe_ridge.predict(X_test) 
```

```python
r2_score(y_test,y_pred_ridge)
```

# using LASSO REGRESSION

```python
# Lasso pipeline
pipe_lasso = Pipeline([
    ('preprocessor', preprocessor),
    ('model', Lasso(alpha=0.1, max_iter=10000))
])

# Train
pipe_lasso.fit(X_train, y_train)

```

```python
# Predict
y_pred_lasso = pipe_lasso.predict(X_test)
```

```python
r2_score(y_test,y_pred_lasso)
```

```python
print("NO Regularization", r2_score(y_test,y_pred_lr))
print("Ridge", r2_score(y_test,y_pred_ridge))
print("Lasso", r2_score(y_test,y_pred_lasso))
```
